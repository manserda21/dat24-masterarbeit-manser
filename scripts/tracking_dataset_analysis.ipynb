### Ordner setzten und Struktur anschauen

```python
tracking_base = Path("/data/manser/soccernet/tracking")
tracking2023_base = Path("/data/manser/soccernet/tracking-2023")

print("tracking exists:", tracking_base.exists())
print("tracking-2023 exists:", tracking2023_base.exists())
```

```python
load_dotenv()

password = os.getenv("SOCCERNET_PASSWORD")
if password is None:
    raise ValueError("SOCCERNET_PASSWORD not found in environment variables")

downloader = SoccerNetDownloader(LocalDirectory="/data/manser/soccernet")
downloader.password = password

# Erst nur einen Tracking-Datensatz laden
downloader.downloadDataTask(
    task="tracking",
    split=["train", "test", "challenge"]
)
```
