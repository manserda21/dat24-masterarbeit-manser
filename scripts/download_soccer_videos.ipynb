## Siege, Unentschieden und Niederlagen zählen

```python
import re
from SoccerNet.utils import getListGames


def parse_result_for_real_madrid(game: str) -> str:
    """
    Bestimmt fuer ein LaLiga-Spiel von Real Madrid,
    ob es ein Sieg, Unentschieden oder eine Niederlage war.

    Rueckgabe:
        "win", "draw" oder "loss"
    """

    match_part = game.split("/")[-1].strip()

    # Erwartetes Format, z.B.:
    # 2016-03-05 - 18-00 Real Madrid 7 - 1 Celta Vigo
    pattern = r"^\d{4}-\d{2}-\d{2}\s*-\s*\d{2}-\d{2}\s+(.*?)\s+(\d+)\s*-\s*(\d+)\s+(.*?)$"
    m = re.match(pattern, match_part)

    if not m:
        raise ValueError(f"Unerwartetes Spiel-Format: {game}")

    home_team = m.group(1).strip()
    home_goals = int(m.group(2))
    away_goals = int(m.group(3))
    away_team = m.group(4).strip()

    if home_team == "Real Madrid":
        real_goals = home_goals
        opp_goals = away_goals
    elif away_team == "Real Madrid":
        real_goals = away_goals
        opp_goals = home_goals
    else:
        raise ValueError(f"Real Madrid nicht im Spiel gefunden: {game}")

    if real_goals > opp_goals:
        return "win"
    elif real_goals < opp_goals:
        return "loss"
    else:
        return "draw"


def count_results_for_season(season: str):
    splits = ["train", "valid", "test", "challenge"]

    wins = 0
    draws = 0
    losses = 0

    for split in splits:
        games = getListGames(split=split)

        for game in games:
            if (
                "spain_laliga" in game
                and season in game
                and "Real Madrid" in game
            ):
                result = parse_result_for_real_madrid(game)

                if result == "win":
                    wins += 1
                elif result == "draw":
                    draws += 1
                elif result == "loss":
                    losses += 1

    return wins, draws, losses


seasons = ["2015-2016", "2016-2017"]

for season in seasons:
    wins, draws, losses = count_results_for_season(season)

    print(f"\nReal Madrid La Liga {season}")
    print("-" * 30)
    print(f"Siege: {wins}")
    print(f"Unentschieden: {draws}")
    print(f"Niederlagen: {losses}")
    print(f"Total: {wins + draws + losses}")
```
