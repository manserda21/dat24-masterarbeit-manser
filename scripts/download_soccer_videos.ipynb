#### Liste für alle `Real Madrid` Spiele

```python
# Hole Listen der Spiele per Split
splits = ["train", "valid", "test", "challenge"]
real_madrid_games = []

for split in splits:
    games = getListGames(split=split)
    for game in games:
        if "Real Madrid" in game:
            real_madrid_games.append((split, game))

print("Real Madrid Spiele:")
for split, game in real_madrid_games:
    print(f"[{split}] {game}")
    
print(len(real_madrid_games))
```

#### Liste für alle Spiele aus der Meister-Saion 2016/2017 (spain_laliga)

```python
# Definiere die Splits, in denen gesucht werden soll
splits = ["train", "valid", "test", "challenge"]
real_madrid_2016_2017_games = []

# Durchsuche alle Spiele in allen Splits
for split in splits:
    games = getListGames(split=split)
    for game in games:
        if "Real Madrid" in game and "2016-2017" in game and "spain_laliga" in game:
            real_madrid_2016_2017_games.append((split, game))

# Ausgabe
print("Real Madrid Spiele aus der Saison 2016/2017:")
for split, game in real_madrid_2016_2017_games:
    print(f"[{split}] {game}")

print(f"\nAnzahl der Spiele: {len(real_madrid_2016_2017_games)}")
```
